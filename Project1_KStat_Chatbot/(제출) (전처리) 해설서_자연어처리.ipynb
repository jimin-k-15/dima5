# CSV 파일

```python
import pandas as pd
df = pd.read_csv('./dataset/HS_33_explanation.csv', encoding='utf-8')
```

```python
client = openai.OpenAI(api_key=MY_KEY)
```

```python
ho_ex_sentence = []

for i, h in enumerate(df['ho_ex']):
    response = client.chat.completions.create(
        model='gpt-3.5-turbo',
        messages =[
            {'role':'system', 'content':f"""
        주어진 규착 해설서 문서를 개조식에서 완성된 문장으로 바꿔줘.
        문서내용: '''{h}'''
        한국어로 답변해주고, 문서 내 내용을 빠짐없이 순서대로 바꿔줘.\
        '(4자리 숫자).(2자리 숫자) 명사' 형식은 '(명사)는 (4자리 숫자.2자리 숫자)에 해당한다'의 형태로 바꿔줘."""},
        ]
    )
    ho_ex_sentence.append(response.choices[0].message.content)
```

```python
df['ho_ex_sentence'] = ho_ex_sentence
```
