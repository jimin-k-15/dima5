```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time


Ryu = [str(i).zfill(2) for i in range(78, 98)]


for n in Ryu:
    # 크롬 드라이버 경로 (필요에 따라 수정)
    driver = webdriver.Chrome(service = Service(), options = webdriver.ChromeOptions())  

    # 사이트 접속
    driver.get("https://ems.epost.go.kr/front.EmsSearchHscode.postal")
    time.sleep(3)
    
    # 검색어 입력
    search_box = driver.find_element(By.ID, "keyword")
    search_box.send_keys(n)
    
    # 검색 버튼 클릭
    driver.find_element(By.XPATH, '//*[@id="frm"]/div/ul/li[2]/a[1]').click()
    
    # 응답 후 실행 코드 만들기,,,
    
    # hs_list = []
    # korean_list = []
    # english_list = []
    results = []
    i = 1
    
    
    while True:
        total = driver.find_element(By.XPATH, '//*[@id="srchResult"]')
        # print(total)
        # print(i)
    
        # tr들을 tbody에서 가져오기
        rows = total.find_elements(By.TAG_NAME, "tr")
        
        for row in rows:
            # 각 tr 내의 td[id=asdf] 3개 가져오기
            tds = row.find_elements(By.TAG_NAME, 'td')
            # print(tds[0].text)
            hs = tds[0].text
            korean = tds[1].text
            english = tds[2].text
            results.append({
                    "HS_CD": hs,
                    "KOR_NAME": korean,
                    "ENG_NAME": english
                })
    
        # print(result)
        # time.sleep(10)
        
        try:
            if i != 1:
                next_btn = driver.find_element(By.XPATH, f'//*[@id="pageItemTd"]/tbody/tr/td[2]/a[{i}]')
                next_btn.click()
                time.sleep(3)
                i += 1
    
            if i == 1:
                next_btn = driver.find_element(By.XPATH, f'//*[@id="pageItemTd"]/tbody/tr/td[2]/a[{i}]')
                next_btn.click()
                time.sleep(3)
                i += 3
        except:
            break
    
    # 브라우저 종료
    driver.quit()
    
    # 데이터프레임으로 저장
    df = pd.DataFrame(results)
    
    # CSV로 저장
    df.to_csv(f'./ems/ems_hscode_{n}_result.csv', index=False, encoding="utf-8-sig")
    
    print(f'{n}류 크롤링 완료! 총', len(df), "건 수집됨.")
    time.sleep(3)
```
