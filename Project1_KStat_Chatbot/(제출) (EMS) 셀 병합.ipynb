```python
import pandas as pd 
ems = pd.read_csv('./datasets/ems_hscode.csv')
ems['HS_CD'] = ems['HS_CD'].astype(str)
ems[ems['HS_CD'].str.startswith('33')]  ## 353개 
```

```python
import pandas as pd
import glob

# 1. 현재 폴더에서 a_로 시작하는 모든 CSV 파일 찾기
csv_files = glob.glob("./ems/ems_hscode_*.csv")

# 2. 파일들을 읽어서 하나의 리스트에 담기
df_list = [pd.read_csv(file) for file in csv_files]

# 3. 데이터프레임들을 하나로 합치기
merged_df = pd.concat(df_list, ignore_index=True)

# 4. 중복 제거 (모든 값이 같은 행 기준)
merged_df = merged_df.drop_duplicates()

# 결과 확인
# print(merged_df)
```

```python
merged_df.reset_index(inplace=True)
```

```python
# merged_df
merged_df = merged_df.sort_values(by='HS_CD')
```

```python
# merged_df = merged_df.drop(columns=['index'])
merged_df.to_csv('./datasets/ems.csv', index=False)
```
